```python
import pandas as pd

# 請求書情報のシート構成案（1レコード = 1請求書）
invoice_main_columns = [
    "請求書ID", "請求元法人名", "請求元郵便番号", "請求元住所",
    "請求先法人名", "請求先氏名", "請求先郵便番号", "請求先住所",
    "発行日", "伝票番号", "振込先", "支払期日", "合計金額"
]

# 商品情報のシート構成案（1請求書に複数行対応）
product_columns = [
    "請求書ID", "商品名", "数量", "単価", "金額"
]

# DataFrameを作成（プレビュー用）
invoice_df = pd.DataFrame(columns=invoice_main_columns)
product_df = pd.DataFrame(columns=product_columns)

# 表示（OpenAI専用コードは削除し、通常の表示方法に変更）
print("▼請求書データ構成案")
print(invoice_df.head(0))

print("\n▼商品データ構成案")
print(product_df.head(0))
```
